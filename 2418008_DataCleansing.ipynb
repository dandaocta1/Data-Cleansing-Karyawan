# **Menampikan** **Struktur** **Variabel** **Data** **Karyawan**

```python
import pandas as pd

# 1. Membaca dataset (header=1 karena nama kolom ada di baris ke-2)
file_name = 'Analisis Kinerja dan Penggajian Karyawan Perusahaan.xlsx'
df = pd.read_excel(file_name, header=1)

# 2. Menampilkan Dimensi Data (Jumlah Baris & Kolom)
print(f"=== DIMENSI DATASET ===")
print(f"Jumlah Baris : {df.shape[0]}")
print(f"Jumlah Kolom : {df.shape[1]}\n")

# 3. Menampilkan Ringkasan Tipe Data dan Missing Value dasar
print("=== INFORMASI TIPE DATA & MISSING VALUE (df.info()) ===")
df.info()

# 4. Membuat Tabel Ringkasan Struktur Variabel yang Lebih Rapi
struktur_df = pd.DataFrame({
    'Nama Kolom': df.columns,
    'Tipe Data': df.dtypes.values,
    'Jumlah Non-Null': df.notnull().sum().values,
    'Jumlah Missing': df.isnull().sum().values,
    'Jumlah Nilai Unik': df.nunique().values,
    'Contoh Nilai': [df[col].dropna().iloc[0] if not df[col].dropna().empty else None for col in df.columns]
})

print("\n=== RINGKASAN STRUKTUR VARIABEL ===")
display(struktur_df)
```

# **Data** **Cleansing** - **Pembersihan** **Header** **&** **Nama** **Kolom**

Mengeliminasi spasi tersembunyi pada nama kolom, menghapus kolom No. yang tidak terpakai, dan merapikan nama Rating Kerja menjadi Rating Kerja

```python
# Merapikan spasi pada nama kolom
df.columns = df.columns.str.strip()

# Hapus kolom 'No.' jika ada
if 'No.' in df.columns:
    df = df.drop(columns=['No.'])

# Rename typo nama kolom RANTING KERJA -> RATING KERJA
df = df.rename(columns={'RANTING KERJA': 'RATING KERJA'})

print("Nama Kolom Setelah Dirapikan:")
print(df.columns.tolist())
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **ID** **Karyawan**

Mengatasi ID ganda/duplikat dan format rusak seperti E00- dengan membentuk ID baru yang unik, rapi, dan konsisten dari E001 hingga E040

```python
# Memperbaiki ID Karyawan agar Unik dan Konsisten (E001 - E040)
df['ID KARYAWAN'] = [f'E{i+1:03d}' for i in range(len(df))]

print("Hasil Standarisasi ID Karyawan (10 Data Pertama):")
display(df[['ID KARYAWAN', 'NAMA']].head(40))
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **Nama**

Tahap ini dilakukan untuk merapikan format penulisan nama karyawan. Pembersihan meliputi penghapusan spasi liar di awal/akhir teks trimming serta pengubahan format huruf menjadi Title Case huruf kapital di setiap awal kata agar tampilan data rapi dan konsisten

```python
# Pengaturan agar Colab menampilkan seluruh baris tanpa terpotong
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

print("=== TOTAL DATASET BERSIH (40 BARIS KARYAWAN) ===")
display(df)
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **Departemen**

 Bagian ini mengatasi inkonsistensi data kategorikal pada kolom DEPARTEMEN. Pembersihan dilakukan dengan menyelaraskan variasi huruf case sensitivit, memetakan istilah/sinonim bahasa Finance & Keuangan rightarrow Finance, Marketing & Pemasaran $\rightarrow$ Marketing, serta mengoreksi typo HDR, Kuangan menjadi 4 departemen utama IT, HRD, Finance, Marketing.

```python
def clean_dept(val):
    val_str = str(val).strip().upper()
    if val_str in ['IT']: return 'IT'
    elif val_str in ['HRD', 'HR', 'HDR']: return 'HRD'
    elif val_str in ['FINANCE', 'KEUANGAN', 'KUANGAN']: return 'Finance'
    elif val_str in ['MARKETING', 'PEMASARAN', 'MAERKETING']: return 'Marketing'
    else: return 'Lainnya'

df['DEPARTEMEN'] = df['DEPARTEMEN'].apply(clean_dept)
display(df[['ID KARYAWAN', 'NAMA', 'DEPARTEMEN']])
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **Usia**

engonversi teks angka ke numerik 'dua puluh' 20, membuang karakter pengotor '?', '2-', menyaring nilai outlier di luar batas usia produktif wajar 18–65 tahun, dan mengimputasi nilai kosong dengan nilai Median usia

```python
def clean_usia(val):
    val_str = str(val).strip().lower()
    if val_str in ['dua puluh', '20']:
        return 20.0
    elif val_str in ['dua lima', '25']:
        return 25.0
    elif val_str in ['?', '2-', 'nan', 'none']:
        return np.nan

    try:
        num = float(val_str)
        if num < 18 or num > 65:
            return np.nan
        return num
    except:
        return np.nan

df['USIA'] = df['USIA'].apply(clean_usia)

# Imputasi dengan Median Usia
median_usia = df['USIA'].median()
df['USIA'] = df['USIA'].fillna(median_usia).astype(int)

print("=== HASIL STANDARISASI USIA (40 BARIS) ===")
display(df[['ID KARYAWAN', 'NAMA', 'USIA']])
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **Gaji**

Membersihkan karakter non-numerik 'ss-', 'ho' dan mengatasi outlier ekstrem. Nilai gaji di luar rentang wajar diimputasi menggunakan Median Gaji berdasarkan Departemen masing-masing.

```python
# 1. Fungsi penyaringan gaji (Rentang Rp 1.000.000 s/d Rp 40.000.000)
def clean_gaji_40m(val):
  val_str = str(val).strip().lower()
  try:
    num = float(val_str)
    # Saring nilai di luar 1 Juta s/d 40 Juta menjadi NaN
    if num < 1000000 or num > 40000000:
      return np.nan
    return num
  except:
    # Mengubah string/huruf/simbol (ss-, ho, Nan) menjadi NaN
    return np.nan


# 2. Terapkan pembersihan awal ke kolom GAJI
df['GAJI'] = df['GAJI'].apply(clean_gaji_40m)

# 3. Imputasi nilai NaN dengan Median Gaji per Departemen
group_medians = df.groupby('DEPARTEMEN')['GAJI'].transform('median')
df['GAJI'] = df['GAJI'].fillna(group_medians)

# 4. Jika masih ada NaN, isi dengan Median Keseluruhan
overall_median = df['GAJI'].median()
df['GAJI'] = df['GAJI'].fillna(overall_median)

# Tampilkan seluruh 40 data kolom GAJI yang sudah bersih
print("=== HASIL CLEANSING GAJI (BATAS MAKSIMAL 40 JUTA) ===")
display(df[['ID KARYAWAN', 'NAMA', 'DEPARTEMEN', 'GAJI']])
```

# **Data** **Cleansing** - **Standarisasi** **Kolom** **Rating** **Kerja**

Memetakan ulang seluruh variasi typo penulisan rating kerja ke dalam 6 skala standar Sangat Baik, Baik, Cukup Baik, Cukup, Kurang, Sangat Buruk dan mengisi nilai yang hilang dengan nilai Modus kategori terbanyak.

```python
def clean_rating(val):
    val_str = str(val).strip().lower()
    if 'sangat baik' in val_str or 'sangar baik' in val_str or 'sanagt baik' in val_str:
        return 'Sangat Baik'
    elif 'cukup baik' in val_str or 'cukup biak' in val_str:
        return 'Cukup Baik'
    elif 'baik' in val_str or 'baikk' in val_str or 'buaikk' in val_str:
        return 'Baik'
    elif 'cukup' in val_str or 'cikip' in val_str:
        return 'Cukup'
    elif 'sangat buruk' in val_str:
        return 'Sangat Buruk'
    elif 'kurang' in val_str or 'kureng' in val_str or 'sangar kurang' in val_str:
        return 'Kurang'
    else:
        return np.nan

df['RATING KERJA'] = df['RATING KERJA'].apply(clean_rating)
df['RATING KERJA'] = df['RATING KERJA'].fillna(df['RATING KERJA'].mode()[0])

print("=== HASIL STANDARISASI RATING KERJA (40 BARIS) ===")
display(df[['ID KARYAWAN', 'NAMA', 'RATING KERJA']])
```

# **Data** **Cleansing** - **Deduplikasi** **&** **Pengecekan** **Missing** **Value**

Mengecek serta menghapus baris duplikat jika ada, lalu memverifikasi bahwa seluruh kolom sudah bebas dari nilai kosong missing value/NaN.

```python
# Hapus baris duplikat
df = df.drop_duplicates()

# Pengecekan akhir nilai missing
print("=== PENGECEKAN NILAI MISSING PER KOLOM ===")
print(df.isnull().sum())
```

# **Data** **Enrichment** **&** **Export** **Dataset** **Akhir**

Pada tahap ini dilakukan penambahan variabel/fitur turunan feature engineering yang relevan dengan domain analisis kinerja dan penggajian karyawan. Penambahan ini bertujuan untuk memperkaya informasi dataset tanpa mengubah, mengelompokkan, atau merusak variabel asli Usia dan Gaji.

```python
# 1. Konversi Rating Kerja ke Skor Numerik (0 - 5)
rating_map = {
    'Sangat Baik': 5,
    'Baik': 4,
    'Cukup Baik': 3,
    'Cukup': 2,
    'Kurang': 1,
    'Sangat Buruk': 0,
}
df['SKOR_RATING'] = df['RATING KERJA'].map(rating_map)

# 2. Perhitungan Akumulasi Gaji Tahunan (12 Bulan)
df['GAJI_TAHUNAN'] = df['GAJI'] * 12

# 3. Selisih Gaji Karyawan Dibanding Rata-rata Departemennya
df['RATA_GAJI_DEPT'] = df.groupby('DEPARTEMEN')['GAJI'].transform('mean')
df['SELISIH_GAJI_DEPT'] = (df['GAJI'] - df['RATA_GAJI_DEPT']).round(0)

# Hapus kolom bantu rata-rata agar dataset tetap rapi
df = df.drop(columns=['RATA_GAJI_DEPT'])

# Pengaturan Tampilan Pandas (Menampilkan Seluruh 40 Baris)
pd.set_option('display.max_rows', 50)

print("=== HASIL AKHIR DATASET KARYAWAN SETELAH DATA ENRICHMENT (40 BARIS) ===")
display(df)

# Simpan Ke File Excel Bersih
df.to_excel('Data_Karyawan_Cleaned.xlsx', index=False)
print("\n✅ File dataset bersih berhasil disimpan: 'Data_Karyawan_Cleaned.xlsx'")
```
